# 🌐 Interactive HTML Visuals Viewer (JupyterLab Desktop)

This notebook displays all 4 generated interactive HTML reports **100% inside the JupyterLab Desktop interface**.

- **PyVis Physics Graph**: Real-time ForceAtlas2 physics network simulation.
- **Plotly 2D Network**: Interactive entity convergence graph.
- **Hype Distribution**: Information pollution spectrum histogram.
- **Outlet Comparison**: Media outlets ranked by BS / hype intensity.

```python
# Ensure project root is in sys.path
import os, sys
from pathlib import Path
NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == 'notebooks' else NOTEBOOK_DIR
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

import ipywidgets as widgets
from IPython.display import display
from src.config import FIGURES_DIR
from src.viz import display_html_in_notebook

print('[OK] Initializing interactive tabs inside JupyterLab Desktop...')

out1 = widgets.Output()
with out1:
    display(display_html_in_notebook(FIGURES_DIR / 'echo_chamber_graph.html', height=680))

out2 = widgets.Output()
with out2:
    display(display_html_in_notebook(FIGURES_DIR / 'network_plotly.html', height=680))

out3 = widgets.Output()
with out3:
    display(display_html_in_notebook(FIGURES_DIR / 'hype_distribution.html', height=680))

out4 = widgets.Output()
with out4:
    display(display_html_in_notebook(FIGURES_DIR / 'outlet_comparison.html', height=680))

tabs = widgets.Tab(children=[out1, out2, out3, out4])
tabs.set_title(0, '🕸️ PyVis Physics Graph')
tabs.set_title(1, '📊 Plotly 2D Network')
tabs.set_title(2, '📈 Hype Distribution')
tabs.set_title(3, '🏛️ Outlet Comparison')
display(tabs)
```
